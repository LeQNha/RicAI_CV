```python
from ultralytics import YOLO
import torch

# Check GPU
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
```

```python
# Dùng YOLOv8n (nano - nhỏ nhất, nhanh nhất)
model = YOLO('yolov8n.pt')

print("✅ Model loaded")
```

```python
results = model.train(
    data='../data/yolo_dataset/data.yaml',
    epochs=50,              # Số epochs
    imgsz=640,              # Image size
    batch=16,               # Batch size (điều chỉnh theo GPU)
    name='rice_leaf_detector',
    patience=10,            # Early stopping
    save=True,
    plots=True,
    
    # ⭐ Fine-tuning params
    lr0=0.001,              # Initial learning rate (thấp hơn train from scratch)
    lrf=0.0001,             # Final learning rate
    warmup_epochs=3,
    
    # Augmentation
    hsv_h=0.015,            # Hue augmentation
    hsv_s=0.7,              # Saturation
    hsv_v=0.4,              # Value
    degrees=10,             # Rotation
    translate=0.1,
    scale=0.5,
    flipud=0.5,             # Vertical flip
    fliplr=0.5,             # Horizontal flip
    mosaic=1.0,
)

print("✅ Training complete!")
print(f"Best model: runs/detect/rice_leaf_detector/weights/best.pt")
```

```python
# Load best model
best_model = YOLO('runs/detect/rice_leaf_detector/weights/best.pt')

# Validate
metrics = best_model.val()

print("\n" + "="*60)
print("VALIDATION METRICS")
print("="*60)
print(f"mAP50:     {metrics.box.map50:.3f}")     # Should be > 0.90
print(f"mAP50-95:  {metrics.box.map:.3f}")       # Should be > 0.70
print(f"Precision: {metrics.box.mp:.3f}")
print(f"Recall:    {metrics.box.mr:.3f}")
print("="*60)
```
