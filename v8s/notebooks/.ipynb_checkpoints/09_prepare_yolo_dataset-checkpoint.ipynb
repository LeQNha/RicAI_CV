```python
import os
import shutil
import random
from pathlib import Path
```

```python
LABELING_DIR = "../data/yolo_labeling"
YOLO_DATASET_DIR = "../data/yolo_dataset"

# YOLOv8 dataset structure:
# yolo_dataset/
#   ├── images/
#   │   ├── train/
#   │   └── val/
#   └── labels/
#       ├── train/
#       └── val/
```

```python
for split in ['train', 'val']:
    os.makedirs(os.path.join(YOLO_DATASET_DIR, 'images', split), exist_ok=True)
    os.makedirs(os.path.join(YOLO_DATASET_DIR, 'labels', split), exist_ok=True)

print("✅ Folders created")
```

```python
images = [f for f in os.listdir(LABELING_DIR) if f.endswith('.jpg')]
random.shuffle(images)

split_idx = int(len(images) * 0.8)
train_images = images[:split_idx]
val_images = images[split_idx:]

print(f"Train: {len(train_images)}")
print(f"Val:   {len(val_images)}")
```

```python
def copy_to_split(image_list, split):
    for img_name in image_list:
        # Copy image
        src_img = os.path.join(LABELING_DIR, img_name)
        dst_img = os.path.join(YOLO_DATASET_DIR, 'images', split, img_name)
        shutil.copy(src_img, dst_img)
        
        # Copy label
        label_name = img_name.replace('.jpg', '.txt')
        src_label = os.path.join(LABELING_DIR, label_name)
        dst_label = os.path.join(YOLO_DATASET_DIR, 'labels', split, label_name)
        
        if os.path.exists(src_label):
            shutil.copy(src_label, dst_label)
        else:
            print(f"⚠️ Missing label: {label_name}")

copy_to_split(train_images, 'train')
copy_to_split(val_images, 'val')

print("✅ Files copied")
```

```python
yaml_content = f"""
# YOLOv8 dataset config
path: {os.path.abspath(YOLO_DATASET_DIR)}
train: images/train
val: images/val

# Classes
nc: 1  # number of classes
names: ['rice_leaf']  # class names
"""

yaml_path = os.path.join(YOLO_DATASET_DIR, 'data.yaml')
with open(yaml_path, 'w') as f:
    f.write(yaml_content)

print(f"✅ Created {yaml_path}")
print("\n" + "="*60)
print("DATASET READY FOR TRAINING")
print("="*60)
print(f"Location: {YOLO_DATASET_DIR}")
print(f"Config:   {yaml_path}")
```
