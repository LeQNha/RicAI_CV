```python
import os
import shutil
import random
from pathlib import Path
```

```python
PROCESSED_DIR = "../data/processed"
LABELING_DIR = "../data/yolo_labeling"
IMAGES_TO_LABEL = 800  # Số ảnh cần label
IMAGES_PER_CLASS = IMAGES_TO_LABEL // 11  # ~73 ảnh/class
```

```python
os.makedirs(LABELING_DIR, exist_ok=True)

sampled_images = []

for class_name in os.listdir(PROCESSED_DIR):
    class_dir = os.path.join(PROCESSED_DIR, class_name)
    if not os.path.isdir(class_dir):
        continue
    
    images = os.listdir(class_dir)
    random.shuffle(images)
    
    # Lấy ~73 ảnh/class
    selected = images[:IMAGES_PER_CLASS]
    
    for img_name in selected:
        src = os.path.join(class_dir, img_name)
        # Đổi tên để tránh trùng: class_originalname.jpg
        new_name = f"{class_name}_{img_name}"
        dst = os.path.join(LABELING_DIR, new_name)
        shutil.copy(src, dst)
        sampled_images.append(new_name)

print(f"✅ Đã copy {len(sampled_images)} ảnh vào {LABELING_DIR}")
```

```python
with open(os.path.join(LABELING_DIR, "file_list.txt"), "w") as f:
    for img_name in sampled_images:
        f.write(img_name + "\n")

print("✅ Tạo file_list.txt xong")
print(f"\n📁 Folder cần label: {LABELING_DIR}")
print(f"📝 Số ảnh: {len(sampled_images)}")
```
