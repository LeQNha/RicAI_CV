```python
import os
import cv2
import matplotlib.pyplot as plt

LABELING_DIR = "../data/yolo_labeling"

# Count labeled images
images = [f for f in os.listdir(LABELING_DIR) if f.endswith('.jpg')]
labels = [f for f in os.listdir(LABELING_DIR) if f.endswith('.txt')]

print(f"Images: {len(images)}")
print(f"Labels: {len(labels)}")
print(f"Coverage: {len(labels)/len(images)*100:.1f}%")
```

```python
def draw_yolo_box(image_path, label_path):
    img = cv2.imread(image_path)
    h, w = img.shape[:2]
    
    with open(label_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 5:
                continue
            
            _, cx, cy, bw, bh = map(float, parts)
            
            # Convert normalized → pixel coords
            x1 = int((cx - bw/2) * w)
            y1 = int((cy - bh/2) * h)
            x2 = int((cx + bw/2) * w)
            y2 = int((cy + bh/2) * h)
            
            cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
    
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
import random
samples = random.sample(images, 10)

fig, axes = plt.subplots(2, 5, figsize=(15, 6))
axes = axes.flatten()

for i, img_name in enumerate(samples):
    img_path = os.path.join(LABELING_DIR, img_name)
    label_path = os.path.join(LABELING_DIR, img_name.replace('.jpg', '.txt'))
    
    if os.path.exists(label_path):
        img = draw_yolo_box(img_path, label_path)
        axes[i].imshow(img)
        axes[i].set_title("Labeled")
    else:
        axes[i].set_title("NOT LABELED", color='red')
    
    axes[i].axis('off')

plt.tight_layout()
plt.savefig("../models/label_verification.png")
plt.show()

print("✅ Check ../models/label_verification.png")
```
