```python
import os
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array, array_to_img
```

```python
# input_root = r"D:\Python Projects\BackUp\downloaded_data_ric_cv"
# output_root = r"D:\Python Projects\BackUp\downloaded_data_ric_cv_augmented"

input_folder = r"D:\Python Projects\BackUp\downloaded_data_ric_cv\healthy"

# Output riêng cho healthy augment
output_folder = r"D:\Python Projects\BackUp\downloaded_data_ric_cv_augmented\healthy"

os.makedirs(output_root, exist_ok=True)
```

```python
augmentor = ImageDataGenerator(

    # Geometric
    rotation_range=45,
    width_shift_range=0.25,
    height_shift_range=0.25,
    shear_range=0.25,
    zoom_range=0.25,
    horizontal_flip=True,
    vertical_flip=True,

    # Color
    brightness_range=[0.6, 1.4],

    fill_mode='nearest'
)
```

```python
# =========================
# Bao nhiêu ảnh augment cho mỗi ảnh gốc
# =========================
augment_per_image = 3

# =========================
# Đếm ảnh trước augment
# =========================
valid_ext = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

before_count = 0
after_count = 0

print("=" * 50)
print("ĐANG AUGMENT DATASET...")
print("=" * 50)
```

```python
# =========================
# Duyệt từng class folder
# =========================
for class_name in os.listdir(input_root):

    class_input_path = os.path.join(input_root, class_name)

    if not os.path.isdir(class_input_path):
        continue

    class_output_path = os.path.join(output_root, class_name)
    os.makedirs(class_output_path, exist_ok=True)

    image_files = [
        f for f in os.listdir(class_input_path)
        if f.lower().endswith(valid_ext)
    ]

    print(f"\nClass: {class_name}")
    print(f"Số ảnh gốc: {len(image_files)}")

    before_count += len(image_files)

    # =========================
    # Augment từng ảnh
    # =========================
    for img_name in image_files:

        img_path = os.path.join(class_input_path, img_name)

        try:
            # Load image
            img = load_img(img_path)

            # Convert sang array
            x = img_to_array(img)

            # Thêm batch dimension
            x = x.reshape((1,) + x.shape)

            # Tạo ảnh augment
            aug_iter = augmentor.flow(
                x,
                batch_size=1
            )

            base_name = os.path.splitext(img_name)[0]

            for i in range(augment_per_image):

                batch = next(aug_iter)

                aug_img = batch[0].astype("uint8")

                aug_img_pil = array_to_img(aug_img)

                save_name = f"{base_name}_aug_{i+1}.jpg"

                save_path = os.path.join(
                    class_output_path,
                    save_name
                )

                aug_img_pil.save(save_path)

                after_count += 1

        except Exception as e:
            print(f"Lỗi với ảnh {img_name}: {e}")


# # =========================
# # Đếm số lượng
# # =========================
# original_images = [
#     f for f in os.listdir(input_folder)
#     if f.lower().endswith(valid_ext)
# ]

# before_count = len(original_images)

# # Đếm số augment cũ nếu có
# existing_augmented = [
#     f for f in os.listdir(output_folder)
#     if f.lower().endswith(valid_ext)
# ]

# existing_count = len(existing_augmented)

# print("=" * 50)
# print("AUGMENT CHO HEALTHY")
# print("=" * 50)

# print(f"Ảnh gốc healthy        : {before_count}")
# print(f"Ảnh augment đã có      : {existing_count}")

# new_augmented_count = 0

# # =========================
# # Augment
# # =========================
# for img_name in original_images:

#     img_path = os.path.join(input_folder, img_name)

#     try:
#         img = load_img(img_path)

#         x = img_to_array(img)

#         x = x.reshape((1,) + x.shape)

#         aug_iter = augmentor.flow(
#             x,
#             batch_size=1
#         )

#         base_name = os.path.splitext(img_name)[0]

#         for i in range(augment_per_image):

#             batch = next(aug_iter)

#             aug_img = batch[0].astype("uint8")

#             aug_img_pil = array_to_img(aug_img)

#             save_name = f"{base_name}_aug_new_{i+1}.jpg"

#             save_path = os.path.join(
#                 output_folder,
#                 save_name
#             )

#             aug_img_pil.save(save_path)

#             new_augmented_count += 1

#     except Exception as e:
#         print(f"Lỗi với {img_name}: {e}")
```

```python
# =========================
# Tổng kết
# =========================
print("\n" + "=" * 50)
print("HOÀN TẤT")
print("=" * 50)

print(f"Tổng ảnh gốc      : {before_count}")
print(f"Tổng ảnh augment  : {after_count}")
print(f"Tổng sau augment  : {before_count + after_count}")

print(f"\nDataset augment nằm tại:")
print(output_root)
```
