```python
from ultralytics import YOLO
import cv2
import os
import shutil

# Load best model
yolo = YOLO('runs/detect/rice_leaf_detector/weights/best.pt')

print("✅ YOLO loaded")
```

```python
test_dir = "../data/processed/brown_spot"
test_images = os.listdir(test_dir)[:100]  # Test 100 ảnh

detected = 0
for img_name in test_images:
    img_path = os.path.join(test_dir, img_name)
    results = yolo(img_path, verbose=False)
    
    if len(results[0].boxes) > 0:
        detected += 1

print(f"\nDetection rate: {detected}/100 = {detected}%")
print("✅ Should be > 95%")
```

```python
PROCESSED_DIR = "../data/processed"
CROPPED_DIR = "../data/cropped_yolo_finetuned"

# Tạo folders
os.makedirs(CROPPED_DIR, exist_ok=True)

CLASS_MAPPING = {
    "bacterial_blight": "bacterial_blight",
    "barrow_brown_leaf_spot": "barrow_brown_leaf_spot",
    "brown_spot": "brown_spot",
    "healthy": "healthy",
    "leaf_blast": "leaf_blast",
    "leaf_scald": "leaf_scald",
    "leaf_smut": "leaf_smut",
    "neck_blast": "neck_blast",
    "rice_hispa": "rice_hispa",
    "sheath_blight": "sheath_blight",
    "tungro": "tungro"
}

for class_name in CLASS_MAPPING.keys():
    os.makedirs(os.path.join(CROPPED_DIR, class_name), exist_ok=True)
```

```python
def crop_leaf_yolo(image_path, yolo_model):
    img = cv2.imread(image_path)
    if img is None:
        return None, "read_error"
    
    # Detect
    results = yolo_model(img, verbose=False)
    boxes = results[0].boxes
    
    if len(boxes) == 0:
        # Fallback: center crop
        h, w = img.shape[:2]
        margin = int(min(h, w) * 0.1)
        x1, y1 = margin, margin
        x2, y2 = w - margin, h - margin
        cropped = img[y1:y2, x1:x2]
        return cropped, "center_crop"
    
    # Get best detection
    best_box = boxes[0].xyxy[0].cpu().numpy().astype(int)
    x1, y1, x2, y2 = best_box
    
    # Add 5% padding
    h, w = img.shape[:2]
    pad = int(min(x2-x1, y2-y1) * 0.05)
    x1 = max(0, x1 - pad)
    y1 = max(0, y1 - pad)
    x2 = min(w, x2 + pad)
    y2 = min(h, y2 + pad)
    
    cropped = img[y1:y2, x1:x2]
    return cropped, "yolo_crop"
```

```python
total = 0
yolo_success = 0
center_crop = 0

for class_name in CLASS_MAPPING.keys():
    print(f"\nProcessing {class_name}...")
    
    src_dir = os.path.join(PROCESSED_DIR, class_name)
    dst_dir = os.path.join(CROPPED_DIR, class_name)
    
    images = os.listdir(src_dir)
    
    for img_name in images:
        src_path = os.path.join(src_dir, img_name)
        
        cropped, method = crop_leaf_yolo(src_path, yolo)
        
        if cropped is None:
            continue
        
        # Resize
        cropped_resized = cv2.resize(cropped, (224, 224))
        
        # Save
        dst_path = os.path.join(dst_dir, img_name)
        cv2.imwrite(dst_path, cropped_resized)
        
        total += 1
        if method == "yolo_crop":
            yolo_success += 1
        else:
            center_crop += 1
```

```python
print("\n" + "="*60)
print("SUMMARY")
print("="*60)
print(f"Total processed:       {total:,}")
print(f"YOLO detected:         {yolo_success:,} ({yolo_success/total*100:.1f}%)")
print(f"Center crop fallback:  {center_crop:,} ({center_crop/total*100:.1f}%)")
print("="*60)
print("\n✅ EXPECTED: YOLO detected > 95%")
```
