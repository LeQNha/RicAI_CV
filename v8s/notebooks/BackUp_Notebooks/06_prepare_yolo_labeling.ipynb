# 06 - Prepare Images for YOLO Labeling
Sample 800 images from dataset for bounding box annotation

```python
import os
import shutil
import random
from pathlib import Path
```

```python
# Config
PROCESSED_DIR = "../data/processed"
LABELING_DIR = "../data/yolo_labeling"
IMAGES_TO_LABEL = 1134

# Số ảnh mỗi class (800 / 11 = ~73)
IMAGES_PER_CLASS = IMAGES_TO_LABEL // 10
```

```python
# Tạo folder
os.makedirs(LABELING_DIR, exist_ok=True)
print(f" Created {LABELING_DIR}")
```

```python
# Sample ảnh từ mỗi class
random.seed(42)
sampled_images = []
total_sampled = 0

for class_name in os.listdir(PROCESSED_DIR):
    class_dir = os.path.join(PROCESSED_DIR, class_name)
    
    if not os.path.isdir(class_dir):
        continue
    
    # Lấy tất cả ảnh
    images = [f for f in os.listdir(class_dir) if f.endswith(('.jpg', '.jpeg', '.png'))]
    random.shuffle(images)
    
    # Sample ~73 ảnh
    selected = images[:IMAGES_PER_CLASS]
    
    print(f"Sampling {class_name}: {len(selected)} images")
    
    # Copy với tên mới (class_originalname.jpg)
    for img_name in selected:
        src = os.path.join(class_dir, img_name)
        new_name = f"{class_name}_{img_name}"
        dst = os.path.join(LABELING_DIR, new_name)
        
        shutil.copy(src, dst)
        sampled_images.append(new_name)
        total_sampled += 1

print(f"\n Total sampled: {total_sampled} images")
```

```python
# Tạo file list
with open(os.path.join(LABELING_DIR, "file_list.txt"), "w") as f:
    for img_name in sampled_images:
        f.write(img_name + "\n")

print(" Created file_list.txt")
```
